```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully")
```

```python
df = pd.read_csv("../data/ndvi_dataset.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
df = pd.read_csv("../data/ndvi_dataset.csv")
df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
print("Missing values:")
print(df.isnull().sum())
print("Class distribution:")
print(df["label"].value_counts())
ndvi_cols = [
    col for col in df.columns
    if col.startswith("ndvi_")
]

print("Number of NDVI columns:", len(ndvi_cols))
print(ndvi_cols)
df[ndvi_cols].describe()
```

```python
df["label"].value_counts().plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Deforestation Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Samples")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for label in df["label"].unique():

    subset = df[df["label"] == label]

    mean_ndvi = subset[ndvi_cols].mean()

    plt.plot(
        range(len(ndvi_cols)),
        mean_ndvi,
        marker="o",
        label=label
    )

plt.xlabel("NDVI Time Step")
plt.ylabel("Mean NDVI")
plt.title("Mean NDVI Temporal Pattern by Class")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
